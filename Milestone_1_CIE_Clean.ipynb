## 1. Read the dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('SuperStore_Sales_Dataset.csv', encoding='latin-1')
df.head()
```

## 2. Understand the dataset

```python
print(df.shape)
print(df.columns)
df.info()
display(df.describe(include='all'))
```

## 3. Check null values

```python
print(df.isnull().sum())
print("Total null values:", df.isnull().sum().sum())
```

## 4. Data transformation

```python
df.columns = df.columns.str.strip()

if 'Order Date' in df.columns:
    df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

if 'Ship Date' in df.columns:
    df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

if 'Row ID' in df.columns:
    df = df.drop(columns=['Row ID'])

if 'Order Date' in df.columns:
    df['Order Year'] = df['Order Date'].dt.year
    df['Order Month'] = df['Order Date'].dt.month

display(df.head())
print(df.dtypes)
```

## 5. Handle null values using mean imputation

```python
numeric_columns = df.select_dtypes(include='number').columns

for column in numeric_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mean())

print(df.isnull().sum())
```

## 6. Handle remaining null values

```python
categorical_columns = df.select_dtypes(include=['object']).columns

for column in categorical_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mode()[0])

print("Total remaining null values:", df.isnull().sum().sum())
```

## 7. Generate insights

### 7.1 Sales and profit by category

```python
category_summary = df.groupby('Category')[['Sales', 'Profit']].sum()
display(category_summary.sort_values('Sales', ascending=False))
```

### 7.2 Sales by region

```python
region_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
display(region_sales)
```

### 7.3 Monthly sales

```python
monthly_sales = df.groupby('Order Month')['Sales'].sum()
display(monthly_sales)
```

## 8. Visualizations

### 8.1 Sales by category

```python
df.groupby('Category')['Sales'].sum().plot(kind='bar', figsize=(8, 5))
plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 8.2 Sales by region

```python
df.groupby('Region')['Sales'].sum().plot(kind='bar', figsize=(8, 5))
plt.title('Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

### 8.3 Monthly sales trend

```python
monthly_sales.sort_index().plot(kind='line', marker='o', figsize=(9, 5))
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```
